```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function and the domain
f = lambda x: np.sin(x)
a, b = 0, 2 * np.pi # Interval
N = 50 # Number of sample points

# Create a uniform grid
x = np.linspace(a, b, N)
y = f(x)


# Plot the sampled points
plt.plot(x, y, 'o-', label='Sampled Points')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Uniform Sampling of f(x) = sin(x)')
plt.legend()
plt.show()
```

```python
# Step size
h = (b - a) / (N - 1)

# Central Difference Approximation for the derivative
df_dx = (f(x + h) - f(x - h)) / (2 * h)

# Plot the approximation and the true derivative
plt.plot(x, np.cos(x), label='Exact Derivative')
plt.plot(x, df_dx, 'o--', label='Central Difference Approximation')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title("Approximating Derivative of f(x) = sin(x) using Central Difference")
plt.legend()
plt.show()

# Forward Difference Approximation
df_dx_forward = (f(x + h) - f(x)) / h

plt.plot(x, np.cos(x), label='Exact Derivative')
plt.plot(x, df_dx_forward, 'o--', label='Forward Difference Approximation')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title("Approximating Derivative of f(x) = sin(x) using Forward Difference")
plt.legend()
plt.show()

# Backward Difference Approximation
df_dx_backward = (f(x) - f(x - h)) / h

plt.plot(x, np.cos(x), label='Exact Derivative')
plt.plot(x, df_dx_backward, 'o--', label='Backward Difference Approximation')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title("Approximating Derivative of f(x) = sin(x) using Backward Difference")
plt.legend()
plt.show()
```

```python
# 4 subplots comparing cos(x), Forward Difference, Backwards Difference, and Central Difference Methods
plt.figure(figsize=(12, 8))

# 1. Exact derivative: cos(x)
plt.subplot(2, 2, 1)
plt.plot(x, np.cos(x))
plt.title('Exact Derivative: cos(x)')
plt.xlabel('x')
plt.ylabel("f'(x)")

# 2. Forward Difference
plt.subplot(2, 2, 2)
plt.plot(x, np.cos(x), label='Exact')
plt.plot(x, df_dx_forward, 'o--', label='Forward Diff')
plt.title('Forward Difference')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.legend()

# Backward Difference
plt.subplot(2, 2, 3)
plt.plot(x, np.cos(x), label='Exact')
plt.plot(x, df_dx_backward, 'o--', label='Backward Diff')
plt.title('Backward Difference')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.legend()

# Central Difference
plt.subplot(2, 2, 4)
plt.plot(x, np.cos(x), label='Exact')
plt.plot(x, df_dx, 'o--', label='Central Diff')
plt.title('Central Difference')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.legend()

plt.tight_layout()
plt.show()
```
