```python
import numpy as np
import matplotlib.pyplot as plt

# Prevent warnings from cluttering output (ChatGPT)
np.seterr(over='ignore', invalid='ignore')

# Define the ODE
def f(t, y):
    return -1000*y + 3000 - 2000*np.exp(-t)

# Forward Euler method
def forward_euler(f, y0, t0, tf, h):
    t_values = np.arange(t0, tf + h, h)
    y_values = np.zeros_like(t_values, dtype=float)  # force float to avoid crash
    y_values[0] = y0
    
    for i in range(1, len(t_values)):
        t_prev = t_values[i-1]
        y_prev = y_values[i-1]
        y_values[i] = y_prev + h * f(t_prev, y_prev)
        
    return t_values, y_values

# Analytical solution
def y_analytical(t):
    return 3 - 0.998*np.exp(-1000*t) - 2.002*np.exp(-t)

# Parameters
y0 = 1
t0 = 0
tf = 30
step_sizes = [1, 0.01, 0.001, 0.00001, 0.0000001]

# Exact solution (computed once)
t_exact = np.linspace(t0, tf, 10000)
y_exact = y_analytical(t_exact)

# Plot everything on one figure
plt.figure(figsize=(12, 6))

for h in step_sizes:
    t_values, y_forward = forward_euler(f, y0, t0, tf, h)
    plt.plot(t_values, y_forward, label=f'Forward Euler (h={h})')

# Plot analytical solution
plt.plot(t_exact, y_exact, 'k--', linewidth=2, label='Analytical Solution')

plt.xlabel('Time (t)')
plt.ylabel('y(t)')
plt.title('Forward Euler Instability for Stiff ODE')
plt.legend()
plt.grid(True)

# keep plot readable despite blow-up (ChatGPT)
plt.ylim([-10, 10])

plt.show()
```
